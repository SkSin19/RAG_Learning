```python

from langchain_core.documents import Document
import os
```

```python
#page_content is the actual content of the doc while metadata is later used for filtering in the vector db.

doc = Document(
    page_content="This is the main text content created for RAG.",
    metadata={
        "source": "example_source.txt",
        "pages": 1,
        "author": "Saksham",
        "date_created": "2026-08-15"
    }
)
```

```python
os.makedirs("../data/text_files", exist_ok=True)
```

```python
sample_texts={
    "../data/text_files/Quarks_Intro.txt" : 
        """Quarks is a digital marketing company. Their slogan is 'From invisible to inevitable'. They make websites, apps, manage seo (search engine optimization), aso (app store optimization), UI/UX, Business Automation, Social Media Handling all for affordable pricing and quote. They are a small yet efficient group of people who excel in their field of work." """,
    
    "../data/text_files/Quarks_Services.txt" : 
        """Quarks provide many things as services: 
        1) Website development
        2) App development
        3) SEO
        4) ASO
        5) Business automation
        6) UI/UX
        7) Social media handling"""
    
}

for filepath, content in sample_texts.items():
    with open(filepath, 'w', encoding="utf-8") as f:
        f.write(content)
        
print("File content written successfully.")
```

```python
from langchain_community.document_loaders import TextLoader

loader = TextLoader("../data/text_files/Quarks_Intro.txt", encoding="utf-8")
text_document = loader.load()
print(text_document)
```

```python
from langchain_community.document_loaders import DirectoryLoader

dir_loader = DirectoryLoader(
    "../data/text_files",
    show_progress=False,
    loader_cls=TextLoader,
    loader_kwargs={'encoding': 'utf-8'},
)

dir_document = dir_loader.load()
print(dir_document)
```
